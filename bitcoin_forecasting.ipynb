## Step 2: Download BTC-USD Data

Now we download Bitcoin historical price data using the ticker `BTC-USD`.  
We use the last 5 years of daily data.

```python
# Import yfinance in this cell so it can run by itself
import yfinance as yf

# Download Bitcoin data from Yahoo Finance
btc_data = yf.download("BTC-USD", period="5y", interval="1d")

# Display the number of rows and columns in the dataset
btc_data.shape
```

**Explanation:** This cell downloads daily Bitcoin prices for the last 5 years. The output shows the dataset size as `(rows, columns)`.

## Step 3: Show the First Rows

Before analyzing the data, we should look at the first few rows to understand its structure.

```python
# Import yfinance in this cell so it can run by itself
import yfinance as yf

# Download the data if it is not already available
if "btc_data" not in globals():
    btc_data = yf.download("BTC-USD", period="5y", interval="1d")

# Show the first 5 rows of the Bitcoin dataset
btc_data.head()
```

**Explanation:** The first rows show the available columns, such as `Open`, `High`, `Low`, `Close`, and `Volume`. For this project, the most important column is `Close` because it represents the final Bitcoin price for each day.

## Step 4: Plot the Bitcoin Close Price

A line chart helps us visually understand how Bitcoin's closing price changed over time.

```python
# Import the libraries in this cell so it can run by itself
import yfinance as yf
import matplotlib.pyplot as plt

# Download the data if it is not already available
if "btc_data" not in globals():
    btc_data = yf.download("BTC-USD", period="5y", interval="1d")

# Create a simple line plot of the Bitcoin closing price
plt.figure(figsize=(12, 6))
plt.plot(btc_data.index, btc_data["Close"], color="blue")
plt.title("Bitcoin Daily Close Price (BTC-USD)")
plt.xlabel("Date")
plt.ylabel("Close Price in USD")
plt.grid(True)
plt.show()
```

**Explanation:** This chart shows the daily closing price of Bitcoin over the last 5 years. It helps us see the overall price movement before we continue with data cleaning, time series analysis, and forecasting in later sections.
